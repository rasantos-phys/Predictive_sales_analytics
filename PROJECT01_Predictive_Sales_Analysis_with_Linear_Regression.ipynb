```python
# Fictitious data
data = {
    'Product': ['T', 'P', 'S', 'M', 'H', 'TBM', 'CB'],
    'Amount': [10, 5, 3, 7, 12, 70, 5],
    'Value': [50, 60, 150, 80, 20, 60, 50]
}

# Convert the dictionary into a DataFrame
df = pd.DataFrame(data)

# View the created DataFrame
print("created DataFrame:")
print(df)
```

```python
# Create simple graphs using matplotlib and seaborn.
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Simple Forecasting with Linear Regression
# Objective: Use linear regression to predict future values

# Step 1: Create a dummy dataset for forecasting

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
# Create a fictitious dataset.
np.random.seed(42)  # To reproduce results
days = np.arange(1, 31).reshape(-1, 1)  # Days from 1 to 30
values = 2 * days + np.random.randn(30, 1) * 2  # Linear relationship with noise
```

```python
# Split the data into training and test data.
X_train, X_test, y_train, y_test = train_test_split(days, values, test_size=0.2, random_state=42)
```

```python
# Training the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Making predictions
predictions = model.predict(X_test)
```

```python
# Display coefficients and intercept
print(f"Coefficient: {model.coef_[0][0]:.2f}")
print(f"Intercept: {model.intercept_[0]:.2f}")
```

```python
# Plot the results
plt.scatter(X_train, y_train, color='blue', label='Training Data')
plt.scatter(X_test, y_test, color='green', label='testing Data')
plt.plot(X_test, predictions, color='red', linewidth=2, label='Linear Regression')
plt.title('Sales Forecast')
plt.xlabel('Days')
plt.ylabel('Values')
plt.legend()
plt.show()
```

```python
# Calculating the R² (coefficient of determination) to numerically measure how well the model fits.
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, predictions)
print(f"R² Score: {r2:.2f}")
```
